## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.

- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

### Cleaning Data

#### Fixing ... problem

```python
# Fixing Data Types
day_df["dteday"] = pd.to_datetime(day_df["dteday"])
hour_df["dteday"] = pd.to_datetime(hour_df["dteday"])
```

```python
# Drop Column
day_df.drop(columns=["instant"], inplace=True)
hour_df.drop(columns=["instant"], inplace=True)
```

```python
# Rename Columns
day_df.rename(columns={"yr": "year", "mnth": "month", "hum": "humidity", "cnt": "total_count"}, inplace=True)
hour_df.rename(columns={"yr": "year", "mnth": "month", "hr": "hour", "hum": "humidity", "cnt": "total_count"}, inplace=True)
```

```python
# Labelling Category

season_map = {1: 'Spring',
              2: 'Summer',
              3: 'Fall',
              4: 'Winter'}
year_map = {0: 2011,
            1: 2012}
month_map = {
    1: 'Jan', 2: 'Feb', 3: 'Mar', 4: 'Apr', 5: 'May', 6: 'Jun',
    7: 'Jul', 8: 'Aug', 9: 'Sep', 10: 'Oct', 11: 'Nov', 12: 'Dec'}
holiday_map = {0: 'Not Holiday',
               1: 'Holiday'}
weekday_map = {
    0: 'Sunday', 1: 'Monday', 2: 'Tuesday', 3: 'Wednesday',
    4: 'Thursday', 5: 'Friday', 6: 'Saturday'}
workingday_map = {0: 'Weekend/Holiday',
                  1: 'Working Day'}
weathersit_map = {
    1: 'Clear/Partly Cloudy',
    2: 'Mist/Cloudy',
    3: 'Light Snow/Rain',
    4: 'Heavy Rain/Snow'}

month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
season_order = ['Spring', 'Summer', 'Fall', 'Winter']
weekday_order = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']

for df in [day_df, hour_df]:
    df['season'] = pd.Categorical(df['season'].map(season_map), categories=season_order, ordered=True)
    df['year'] = df['year'].map(year_map).astype('category')
    df['month'] = pd.Categorical(df['month'].map(month_map), categories=month_order, ordered=True)
    df['holiday'] = df['holiday'].map(holiday_map).astype('category')
    df['weekday'] = pd.Categorical(df['weekday'].map(weekday_map), categories=weekday_order, ordered=True)
    df['workingday'] = df['workingday'].map(workingday_map).astype('category')
    df['weathersit'] = df['weathersit'].map(weathersit_map).astype('category')
```

```python
# Check Cleaning Data

print("=============  day_df cleaned =============")
day_df.info()
display(day_df.head(2))
print("\n============= hour_df cleaned ============")
hour_df.info()
display(hour_df.head(2))
```

```python
# Save Cleaned Data
# Menyisipkan informasi cuaca harian dari day_df ke hour_df agar agregasi harian di dashboard 100% konsisten
daily_weather_map = day_df.set_index('dteday')['weathersit']
hour_df['day_weathersit'] = hour_df['dteday'].map(daily_weather_map)

# Mengekspor dataset bersih menjadi main_data.csv untuk kebutuhan dashboard Streamlit
hour_df.to_csv("main_data.csv", index=False)
print("File main_data.csv berhasil disimpan!")
```

    - `day_df` --> 15 kolom`
    - `hour_df` --> 16 kolom`
  - serta penamaan kolom jadi lebih jelas (`year`, `month`, `hour`, `humidity`, dan `total_count`).

- **Tipe Data Sudah Benar:** Kolom `dteday` berhasil dikonversi yang awalnya

    `object` --> `datetime64[ns]`
    
    sedangkan variabel kategorikal (`season`, `year`, `month`, `holiday`, `weekday`, `workingday`, dan `weathersit`) telah diubah menjadi tipe `category` dengan urutan kronologis pada `season`, `month`, dan `weekday`.

- **Sudah bisa untuk EDA:** Seluruh nilai kategori sudah ada labelnya, sudah dapat di grouping dan lanjutan untuk membuat grafik-grafiknya untuk menjawab pertanyaan SMART.

## Exploratory Data Analysis (EDA)

### Explore ...

```python
# 1. Rata-rata penyewaan harian berdasarkan Musim (season)
season_eda = day_df.groupby('season', observed=True).agg(
    jumlah_hari=('total_count', 'count'),
    rata_rata_harian=('total_count', 'mean'),
    rata_rata_casual=('casual', 'mean'),
    rata_rata_registered=('registered', 'mean'),
    total_penyewaan=('total_count', 'sum')
).round(2)

print("=== RATA-RATA PENYEWAAN HARIAN BERDASARKAN MUSIM ===")
display(season_eda)

# 2. Rata-rata penyewaan harian berdasarkan Kondisi Cuaca (weathersit)
weather_eda = day_df.groupby('weathersit', observed=True).agg(
    jumlah_hari=('total_count', 'count'),
    rata_rata_harian=('total_count', 'mean'),
    rata_rata_casual=('casual', 'mean'),
    rata_rata_registered=('registered', 'mean'),
    total_penyewaan=('total_count', 'sum')
).round(2)

print("\n=== RATA-RATA PENYEWAAN HARIAN BERDASARKAN KONDISI CUACA ===")
display(weather_eda)

# 3. Kombinasi Musim dan Kondisi Cuaca (Pivot Table)
season_weather_pivot = day_df.pivot_table(
    index='season',
    columns='weathersit',
    values='total_count',
    aggfunc='mean',
    observed=True
).round(2)

print("\n=== RATA-RATA PENYEWAAN HARIAN: MUSIM VS CUACA ===")
display(season_weather_pivot)
```

```python
# 1. Rata-rata penyewaan per jam berdasarkan tipe hari (workingday)
hourly_pattern = hour_df.groupby(['workingday', 'hour'], observed=True).agg(
    avg_casual=('casual', 'mean'),
    avg_registered=('registered', 'mean'),
    avg_total=('total_count', 'mean')
).round(2).reset_index()

print("=== SAMPEL POLA PENYEWAAN PER JAM (WORKING DAY VS WEEKEND/HOLIDAY) ===")
display(hourly_pattern.head(10))

# 2. Mencari 3 jam puncak tertinggi untuk masing-masing segmen pengguna dan tipe hari
for day_type in ['Working Day', 'Weekend/Holiday']:
    subset = hourly_pattern[hourly_pattern['workingday'] == day_type]
    top_casual = subset.nlargest(3, 'avg_casual')[['hour', 'avg_casual']]
    top_reg = subset.nlargest(3, 'avg_registered')[['hour', 'avg_registered']]

    print(f"\n=== TOP 3 JAM PUNCAK ({day_type.upper()}) ===")
    print("Pengguna Casual:")
    display(top_casual)
    print("Pengguna Registered:")
    display(top_reg)
```

```python
# 1. Agregasi total penyewaan bulanan per tahun
monthly_sales = day_df.groupby(['year', 'month'], observed=True).agg(
    total_rentals=('total_count', 'sum'),
    casual_rentals=('casual', 'sum'),
    registered_rentals=('registered', 'sum')
).reset_index()

# 2. Menghitung pertumbuhan Month-over-Month (MoM %) di dalam masing-masing tahun
monthly_sales['MoM_growth_%'] = (
    monthly_sales.groupby('year', observed=True)['total_rentals']
    .pct_change() * 100
).round(2)

# 3. Menyaring khusus Kuartal 4 (Oct, Nov, Dec) untuk perbandingan 2011 vs 2012
q4_comparison = monthly_sales[monthly_sales['month'].isin(['Oct', 'Nov', 'Dec'])].copy()

print("=== PERBANDINGAN PERTUMBUHAN BULANAN (MoM) Q4: 2011 VS 2012 ===")
display(q4_comparison)

# 4. Ringkasan total volume Q4 2011 vs Q4 2012 (Year-over-Year)
q4_summary = q4_comparison.groupby('year', observed=True)['total_rentals'].sum()
yoy_q4_growth = ((q4_summary[2012] - q4_summary[2011]) / q4_summary[2011]) * 100
print(f"\nTotal Penyewaan Q4 2011: {q4_summary[2011]:,}")
print(f"Total Penyewaan Q4 2012: {q4_summary[2012]:,}")
print(f"Pertumbuhan Q4 YoY (2011 ke 2012): {yoy_q4_growth:.2f}%")
```

**Insight:**
- **Dampak Musim & Cuaca (Pertanyaan 1):** Penyewaan sepeda mencapai rata-rata harian tertinggi pada musim gugur (*Fall*, sekitar 5.644 sepeda/hari) dan terendah pada musim semi (*Spring*, sekitar 2.604 sepeda/hari). Dari sisi cuaca, kondisi cerah/berawan sebagian (*Clear/Partly Cloudy*) menghasilkan rata-rata tertinggi (4.877 sepeda/hari), sedangkan hujan/salju ringan (*Light Snow/Rain*) menurunkan rata-rata penyewaan hingga lebih dari setengahnya (~1.803 sepeda/hari), serta tidak ada hari dengan cuaca ekstrem (*Heavy Rain/Snow*) dalam catatan agregat harian.
- **Perbedaan Pola Jam Pengguna (Pertanyaan 2):** Terdapat kontras perilaku yang sangat jelas antara kedua segmen pengguna. Pada hari kerja (*Working Day*), pengguna terdaftar (*registered*) membentuk pola dua puncak (*bimodal*) yang tajam pada jam berangkat kerja (**08.00**) dan jam pulang kerja (**17.00–18.00**), sementara pengguna kasual (*casual*) relatif rendah. Sebaliknya, pada akhir pekan/hari libur (*Weekend/Holiday*), penyewaan didominasi oleh aktivitas rekreasi siang hari dengan satu puncak landai antara pukul **12.00 hingga 16.00** untuk kedua segmen.
- **Pertumbuhan Kuartal 4 (Pertanyaan 3):** Secara volume keseluruhan, total penyewaan Q4 2012 tumbuh signifikan dibandingkan Q4 2011 (meningkat lebih dari 40% secara *Year-over-Year*). Namun, secara bulanan (*Month-over-Month*), setiap bulan di Q4 (`Oct` ke `Nov` hingga `Dec`) konsisten mengalami kontraksi negatif baik pada 2011 maupun 2012 akibat peralihan suhu menuju musim dingin—dengan penurunan tajam pada akhir Oktober 2012 yang turut dipengaruhi oleh badai Hurricane Sandy di Washington D.C.

## Visualization & Explanatory Analysis

## Analisis Lanjutan (Opsional)

#### Time-of-Day Clustering (Manual Grouping / Binning)
**Tujuan Teknik Analisis:**  
Teknik *manual grouping* (*binning*) digunakan untuk mengelompokkan 24 jam operasional (`00:00–23:59`) ke dalam 4 klaster waktu (*Night*, *Morning*, *Afternoon*, dan *Evening*). Tujuannya adalah menyederhanakan pola distribusi permintaan per jam agar tim operasional dapat mengidentifikasi blok waktu paling padat untuk alokasi armada serta blok waktu paling sepi untuk penjadwalan perawatan (*maintenance*) sepeda.

```python
# Membuat klaster waktu operasional (Time-of-Day Binning) pada hour_df
bins = [-1, 5, 11, 17, 23]
labels = ['Night (00:00-05:59)', 'Morning (06:00-11:59)', 'Afternoon (12:00-17:59)', 'Evening (18:00-23:59)']

hour_df['time_cluster'] = pd.cut(hour_df['hour'], bins=bins, labels=labels)

# Agregasi rata-rata dan total penyewaan berdasarkan klaster waktu dan tipe hari
cluster_analysis = hour_df.groupby(['workingday', 'time_cluster'], observed=True).agg(
    avg_casual=('casual', 'mean'),
    avg_registered=('registered', 'mean'),
    avg_total=('total_count', 'mean'),
    total_rentals=('total_count', 'sum')
).round(2).reset_index()

print("=== ANALISIS KLASTER WAKTU OPERASIONAL ===")
display(cluster_analysis)

# Visualisasi perbandingan klaster waktu pada Hari Kerja vs Akhir Pekan/Libur
plt.figure(figsize=(12, 6))
ax = sns.barplot(
    data=cluster_analysis,
    x='time_cluster',
    y='avg_total',
    hue='workingday',
    palette=['#ff7f0e', '#1f77b4']
)

plt.title('Rata-rata Penyewaan Sepeda per Jam berdasarkan Klaster Waktu Operasional', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Klaster Waktu (Time-of-Day Cluster)', fontsize=11)
plt.ylabel('Rata-rata Penyewaan per Jam (Unit)', fontsize=11)
plt.ylim(0, 360)
plt.legend(title='Tipe Hari')

for container in ax.containers:
    ax.bar_label(container, fmt='%.1f', padding=4, fontsize=9.5, fontweight='bold')

plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- **Dominasi Klaster Siang & Sore:** Secara keseluruhan, klaster **Afternoon (12:00–17:59)** mencatat rata-rata penyewaan per jam tertinggi baik pada hari kerja (**264,1 unit/jam**) maupun akhir pekan/libur (**316,6 unit/jam**), diikuti oleh klaster **Evening (18:00–23:59)** pada hari kerja (**232,1 unit/jam**).
- **Pergeseran Beban Armada Pagi vs Siang:** Pada hari kerja, klaster **Morning (06:00–11:59)** memiliki rata-rata penyewaan yang jauh lebih tinggi (**213,5 unit/jam**) dibandingkan akhir pekan/libur (**149,5 unit/jam**) karena tingginya aktivitas komuter pagi. Sebaliknya, klaster **Night (00:00–05:59)** memiliki permintaan terendah (< 35 unit/jam), sehingga menjadi waktu paling ideal untuk pemeliharaan dan redistribusi sepeda.
